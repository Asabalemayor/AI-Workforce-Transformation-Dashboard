```python
merged_df = (
    fact
    .merge(country, on='country_id', how ='left')
    .merge(date, on='date_id', how ='left')
    .merge(skill, on='skill_category_id', how ='left')
    .merge(industry, on='industry_id', how ='left')
)
```

```python
## How has AI adoption changed quarter over quarter since the Generative AI era began in Q4 2022?
quarterly_ai = (
    merged_df
    .groupby(['year_quarter_label'])['ai_adoption_rate']
    .mean()
    .reset_index()
)
quarterly_ai
```

```python
## The percentage increase from 2022-Q3 to 2022-Q4.

old = 22.481176
new = 40.234583

increase = ((new - old) / old) * 100
print(increase)
```

```python
# Is the rate of job displacement accelerating or stabilizing across 2023–2024?
quarterly_job_displacement = (
    merged_df
    .groupby(['year_quarter_label'])['jobs_displaced_count']
    .mean()
    .reset_index()
)
quarterly_job_displacement
```

```python
# how many records each quarter contains.
merged_df.groupby('year_quarter_label').size()
```

```python
# Which quarters show the steepest rise in ai_tool_usage_hours_per_week?
# Calculate the average AI usage per quarter
quarterly_usage = (
    merged_df
    .groupby('year_quarter_label')['ai_tool_usage_hours_per_week']
    .mean()
    .reset_index()
)
    
quarterly_usage['Rise'] = quarterly_usage['ai_tool_usage_hours_per_week'].diff()

# Return only the quarter with the steepest rise
steepest_rise = quarterly_usage.loc[quarterly_usage['Rise'].idxmax()]

print(steepest_rise)
```

```python
# Is job creation offsetting job displacement?
quarterly_jobs = (
    merged_df
    .groupby('year_quarter_label')[['jobs_displaced_count', 'jobs_created_count']]
    .mean()
    .reset_index()
)
```

```python
quarterly_jobs['job_replacement_ratio'] = (
    quarterly_jobs['jobs_created_count']
    / quarterly_jobs['jobs_displaced_count']
)

quarterly_jobs
```

```python
# Which industries combine high automation susceptibility with low average AI investment (% revenue), suggesting they are under-prepared?
industry_preparedness = (
    merged_df
    .groupby('industry_name')[[
        'automation_susceptibility',
        'avg_ai_investment_pct_revenue'
    ]]
    .mean()
    .reset_index()
)
industry_preparedness
```

```python
industry_preparedness.sort_values(
    by=['automation_susceptibility',
        'avg_ai_investment_pct_revenue'],
    ascending=[False, True]
)
```

```python
# Which skill categories show the largest gap between ai_replaceability_score and median_reskilling_duration_months?
skill_gap = (
    merged_df
    .groupby('skill_category_name')[[
        'ai_replaceability_score',
        'median_reskilling_duration_months'
    ]]
    .mean()
    .reset_index()
)
skill_gap['reskilling_gap'] = (
    skill_gap['ai_replaceability_score']
    - skill_gap['median_reskilling_duration_months']
)

largest_gap = skill_gap.loc[skill_gap['reskilling_gap'].idxmax()]
print(largest_gap)
```

```python
# Are creative and interpersonal skill categories more resilient to displacement than administrative or manual ones?
skill_gap.sort_values(
    by='reskilling_gap',
    ascending=False
)
```

```python
# How does displacement_risk_index correlate with digital_infrastructure_score and stem_graduates_per_100k by country?
country_analysis = (
    merged_df
    .groupby('country_name')[[
        'displacement_risk_index',
        'digital_infrastructure_score',
        'stem_graduates_per_100k'
    ]]
    .mean()
    .reset_index()
)

country_analysis.head()
```

```python
# Calculate the Correlations
country_analysis[[
    'displacement_risk_index',
    'digital_infrastructure_score',
    'stem_graduates_per_100k'
]].corr()
```

```python
# Do countries with more mature ai_policy_maturity show better job creation-to-displacement ratios?
merged_df['job_replacement_ratio'] = (
    merged_df['jobs_created_count']
    / merged_df['jobs_displaced_count']
)
```

```python
# Create the Ratio
merged_df['ai_policy_maturity'].unique()
```

```python
## Group by Policy Maturity
policy_analysis = (
    merged_df
    .groupby('ai_policy_maturity')['job_replacement_ratio']
    .mean()
    .reset_index()
)

policy_analysis
```

```python
# Which regions show the widest gap between developed and emerging economy adoption rates?
regional_gap = (
    merged_df
    .groupby(['region', 'development_tier'])['ai_adoption_rate']
    .mean()
    .unstack()          # Turns Developed and Emerging into separate columns
)

regional_gap['adoption_gap'] = (
    regional_gap['Developed'] - regional_gap['Emerging']
)

regional_gap = regional_gap.sort_values(
    by='adoption_gap',
    ascending=False
)

regional_gap
```

```python
merged_df.groupby(['region', 'development_tier'])['country_name'].nunique()
```

```python
# Is reskilling_investment_usd keeping pace with jobs_displaced_count at an industry level?
industry_reskilling = (
    merged_df
    .groupby('industry_name')[[
        'jobs_displaced_count',
        'reskilling_investment_usd'
    ]]
    .mean()
    .reset_index()
)

industry_reskilling
```

```python
# Investment per displaced worker
industry_reskilling['investment_per_displaced_worker'] = (
    industry_reskilling['reskilling_investment_usd']
    / industry_reskilling['jobs_displaced_count']
)

industry_reskilling.sort_values(
    by='investment_per_displaced_worker'
)
```

```python
# Which country–industry combination has the largest shortfall between displacement risk and reskilling spending?
country_industry = (
    merged_df
    .groupby(['country_name', 'industry_name'])[
        ['displacement_risk_index', 'reskilling_investment_usd']
    ]
    .mean()
    .reset_index()
)

# Convert investment to millions (optional, makes the numbers easier to read)
country_industry['investment_million'] = (
    country_industry['reskilling_investment_usd'] / 1_000_000
)

country_industry['shortfall_score'] = (
    country_industry['displacement_risk_index']
    / country_industry['investment_million']
)

country_industry = country_industry.sort_values(
    by='shortfall_score',
    ascending=False
)

country_industry.head(10)
```
